# Harmonic and Anharmonic Oscillations of a Boat

Simple Harmonic Oscillation (SHO) is everyone's favorite type of periodic motion.  It's simple, easy to understand, and has a known solution.  But SHO is not the only kind of periodic motion.  In this set of exercises we'll investigate an *anharmonic* oscillator. We'll explore characteristics of its motion, and try to get an understanding of the circumstances under which the motion can be approximated as simple harmonic.

## A straight-sided boat

Imagine a boat with a flat bottom and vertical sides floating in the water. There will be two forces on this boat: its weight $mg$, and the buoyant force $-\rho g V$ (taking downward to be the positive $y$ direction). The volume $V$ appearing in the buoyant force is the volume of water displaced by the boat; in other words, the volume of boat that is *below* the water level. In equilibrium, the bottom of the boat will be some distance $y_{\mathrm{EQ}}$ below the water, so $V = Ay_{\mathrm{EQ}}$, where $A$ is the area of the boat's cross-section.

If we then push the boat down some additional distance $y$ and let go, the boat will bob up and down.  According to Newton's second law, we have

$$ \Sigma F = mg - \rho g A \left( y_{\mathrm{EQ}} + y \right) = ma_y .\tag{1}$$

At equilibrium, $y=a_y=0$ and we obtain $mg = \rho g A y_{\mathrm{EQ}}.$  Using this expression to eliminate $y_{\mathrm{EQ}}$ from Equation 1 yields

$$ a_y = \frac{d^2y}{dt^2} = -\frac{\rho g A}{m}y = -\omega^2y,$$
where we have defined $ \omega \equiv \sqrt{\frac{\rho g A}{m}}$.  This is the equation for simple harmonic oscillation with angular frequency $\omega$ and period $ T = 2\pi/\omega.$

The width $w$ of the hull at the waterline depends on the depth of the boat below waterline $h$ according to

$$ w = \beta h ,$$
where $\beta$ is a positive constant.  The volume of water displaced by the boat is the area of the triangle below water level, times the length $L$ of the boat:

$$ V = \frac{1}{2}\beta h^2 L .$$

Replace $h$ with $(y_{\mathrm{EQ}} + y)$ and follow the example in the introduction to show that the boat does *not* undergo simple harmonic motion by obtaining an equation of motion of the form $$ \frac{d^2y}{dt^2} = -\omega_0^2\left( 1 + \frac{y}{2y_{\mathrm{EQ}}} \right) y \tag{2}.$$  In this form, you can see that the motion is approximately that of a SHO with angular frequency $\omega_0$ only if $\frac{y}{2y_{\mathrm{EQ}}}$ is small. What is $\omega_0$?

### Solve the equation of motion

To calculate the motion of the boat, we must solve the 2nd-order ordinary differential equation in Equation 2.  The general procedure for such cases is to recast the single 2nd-order differential equation into a pair of coupled 1st order differential equations.  To do so, we begin by observing that $$\frac{dy}{dt}=v,\tag{3}$$ in terms of which Equation 2 becomes $$\frac{d^2y}{dt^2} = \frac{d}{dt}\frac{dy}{dt}= \frac{dv}{dt} = -\omega_0^2\left( 1 + \frac{y}{2y_{\mathrm{EQ}}} \right) y\tag{4}.$$

Equations 3 and 4 represent a system of two coupled 1st-order differential equations, which we can now solve using the `odeint` function from the `scipy.integrate` module. Documentation on this function can be found [here](https://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.odeint.html).

Define a function `ODEs(state,t)` that takes as argument a two-element list containing the state of the system ($y$ and $v$) and returns a two-element list containing the derivatives calculated according to Equations 3 and 4.  Assume $\omega=1$ Hz and $y_{\mathrm{EQ}}=10$ m.  

This function, along with a two-element list containing the initial state of the system and an array of time values, is all that is needed  to solve the coupled system of differential equations.  Pass this info to `odeint`.  It will solve the equations and return a 2D array where each column represents either $y(t)$ or $v(t)$.
