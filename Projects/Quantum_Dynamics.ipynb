Let's begin by assuming that a particle is initially confined within the well, with an initial wavefunction given by $\Psi (x,0) = Ax\left(1-x\right)\left(\frac{1}{2}-x\right)^2$.  If this wavefunction is normalized properly, the integral of the probability density $\left|\Psi(x,0)\right|^2$ over all space must equal one.  Using $\int_{-\infty}^{\infty}\left|\Psi(x,0)\right|^2 dx=1$, calculate the value of the normalization factor $A$ *analytically*.  

\begin{align}
    \int_{-\infty}^{\infty}\left|\Psi(x,0)\right|^2 dx &= 1\\
    \int_0^1\left|Ax\left(1-x\right)\left(\frac{1}{2}-x\right)\right|^2 dx &= 1\\
    |A|^2\int_0^1 \left(\frac{x}{4}-\frac{5x^2}{4}+2x^3-x^4\right)^2 dx &= 1\\
    |A|^2\times 0.00009921&=1\\
    A=100.4
\end{align}

The set of solutions to the time-independent Schr&#246;dinger equation ("stationary states") for an inifite well with width $L=1$, $\psi_n(x)=\sqrt{2}\sin\left(n\pi x\right)$, constitute a complete orthonormal basis.  This means that any arbitrary wavefunction, including $\Psi(x,0)$, can be expressed as a linear superposition of the stationary states  
\begin{equation}\tag{1}
    \Psi(x,0)=\sum_{n=1}^{\infty} C_n\psi_n(x).
\end{equation}
The coefficicients $C_n$ can be found using the relation
\begin{equation}\tag{2}
    C_n=\int_{-\infty}^{\infty}\Psi(x,0)^*\psi_n(x) dx.
\end{equation}
Using Eqn. 2, calculate these coefficients *numerically* for the stationary states with $1\leq n\leq20$.  NB: keep in mind that python sequences are zero indexed, while the smallest index $n$ appearing in Eqn. 2 is one.

You should observe that all even coefficients are approximately zero, while the magnitude of odd coefficients generally decreases with increasing $n$.  

Therefore, for the remainder of the assignment, we will neglect all terms in Eqn. 1 with even values of $n$ or $n\geq10$, as their contribution to the total wavefunction is negligible.  Plot the wavefunctions for the 5 lowest odd states, weighted by the corresponding coefficient $C_n$.

Sum these weighted stationary states, as in Eqn. 1, and compare the result to the original wavefunction $\Psi(x)$.  You should find that the two distributions agree very well.

The stationary states evolve in time according to
\begin{equation}\tag{3}
    \psi_n(x,t)=e^{-iE_nt/\hbar}\psi_n(x),
\end{equation}
where $E_n=\frac{n^2\pi^2\hbar^2}{2m}$ is the energy of the $n$th stationary state.  For simplicity, let's assume that $\frac{\pi^2\hbar}{2m}=1$, so that $E_n=n^2\hbar$ and Eqn. 3 simplifies to
\begin{equation}\tag{4}
    \psi_n(x,t)=e^{-in^2t}\psi_n(x).
\end{equation}

On 5 separate sets of axes, construct an animation that shows the time evolution of the real and imaginary component of each stationary state.

While the stationary states in Eqn. 4 are purely real, the exponential factor represents a complex phase. This phase causes each of the stationary states to rotate in the complex plane as time evolves, while the shape remains constant.  The rotation rate is determined by $n^2$, so each stationary state rotates at a different rate, which you should see in your animation.  However, note that this complex phase does not affect the probability density associated with a particular stationary state, which is constant in time (hence the name "stationary" to describe these states)
\begin{equation}
    \left|\psi_n(x,t)\right|^2 = \psi_n(x,t)^* \times \psi_n(x,t) = \psi_n(x)^*e^{in^2t}e^{-in^2t}\psi_n(x) = \left|\psi_n(x)\right|^2.
\end{equation}
